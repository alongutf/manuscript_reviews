```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
rho = 0.4
M = rho * np.ones((3,3))
#M[0,2] *= -1
M = np.triu(M)
M = M + M.T
np.fill_diagonal(M,1)
np.linalg.eig(M)[0]
n=1000
X = np.random.multivariate_normal([0,0,0],M,n)
# 2. Create the figure and 3D axes
corr_mat = X.T @ X /n
sns.heatmap(corr_mat, cmap='viridis')
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')


# 3. Plot the data
# 'c' sets the color (can be mapped to Z), 'marker' sets the shape
scatter = ax.scatter(X[:,0], X[:,1], X[:,2], s=50)
```
